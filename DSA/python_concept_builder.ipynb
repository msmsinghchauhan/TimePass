```python
import time
import re

class LogParser:
    def __init__(self,log_file:str|None=None,log_string:str|None=None):
        self.log_file=log_file
        self.log_string=log_string
        self.start_time=0.0
        self.end_time=0.0
        self.elapsed_time=0.0
        self.error_counts=[]

    def __enter__(self):
        self.start_time=time.perf_counter()
        return self

    def __exit__(self,a,b,c):
        self.end_time=time.perf_counter()
        self.elapsed_time=self.end_time-self.start_time
        print(f"\n[Timer Ended] Duration: {self.elapsed_time:.4f} seconds")
        return False

    def parse_logs(self):
        pattern=r"ERROR|CRITICAL|WARNING"
        if self.log_file:
            with open(self.log_file,"r") as f:
                for line in f:
                    if re.search(pattern, line,re.IGNORECASE):
                        self.error_counts.append(line)
        if self.log_string:
            for line in self.log_string.split("\n"):
                if re.search(pattern, line,re.IGNORECASE):
                    self.error_counts.append(line)  


log_string= """2026-06-22 18:10:15 [ERROR] [main] com.example.app.OrderService - Failed to process order #99482
java.lang.NullPointerException: Cannot invoke "String.length()" because "userName" is null
    at com.example.app.OrderService.formatName(OrderService.java:42)
    at com.example.app.OrderService.process(OrderService.java:21)
    at java.base/java.lang.Thread.run(Thread.java:833)"""


with LogParser(log_string=log_string) as lp:
    lp.parse_logs()
    print(f"Found {len(lp.error_counts)} error lines:")
    for error_line in lp.error_counts:
        print(error_line)

```

Usage of decorator to add performance counter


```python
import time
import functools


def perf_timer_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time=time.perf_counter()
        try:
            result=func(*args, **kwargs)
        finally:
            end_time=time.perf_counter()
            elapsed_time=end_time-start_time
        print(f"\n[Timer Ended] Duration: {elapsed_time:.4f} seconds")
        return result
    return wrapper

@perf_timer_decorator
def generate_fibonacci(n):
    fib_sequence = [0,1]
    if n<=0 :
        return []
    elif n==1:
        return [0]
    elif n==2:
        return [0,1]
    else:
        for i in range(2,n):
            fib_sequence.append(fib_sequence[i-1] + fib_sequence[i-2])  
        return fib_sequence
    


generated_fib_sequence=generate_fibonacci(10)
print(generated_fib_sequence)



    
```

```python
import asyncio
import nest_asyncio
import pdb

nest_asyncio.apply()

condition = asyncio.Condition()
shared_data = None

async def consumer():
    global shared_data
    async with condition:
        await condition.wait_for(lambda: shared_data is not None)
        print(f"Consumer got: {shared_data}")

async def producer():
    global shared_data
    await asyncio.sleep(1)
    async with condition:
        shared_data = "Data is ready!"
        condition.notify_all()



# --- The Main Coroutine and Entry Point ---
async def main():
    # asyncio.gather runs both coroutines concurrently
    #pdb.set_trace()  # Set a breakpoint here to inspect the state before running the tasks
    await asyncio.gather(
        consumer(),
        producer()
    )

if __name__ == "__main__":
    # This starts the event loop and executes the main coroutine
    asyncio.run(main())

        
```

```python
import asyncio
import nest_asyncio


nest_asyncio.apply()


class AsyncRWLock:
    def __init__(self):
        self._condition = asyncio.Condition()
        self._readers = 0
        self._writers = 0

    async def acquire_read(self):
        async with self._condition:
            # Wait if a writer is currently active
            await self._condition.wait_for(lambda: self._writers == 0)
            self._readers += 1

    async def release_read(self):
        async with self._condition:
            self._readers -= 1
            if self._readers == 0:
                self._condition.notify_all()

    async def acquire_write(self):
        async with self._condition:
            # Wait if anyone else is reading or writing
            await self._condition.wait_for(lambda: self._readers == 0 and self._writers == 0)
            self._writers = 1

    async def release_write(self):
        async with self._condition:
            self._writers = 0
            self._condition.notify_all()

# --- Simulation Code ---

rw_lock = AsyncRWLock()

async def reader(reader_id):
    for _ in range(2):
        await rw_lock.acquire_read()
        print(f"📖 Reader {reader_id} is READING the file.")
        await asyncio.sleep(0.5)  # Simulate file reading I/O
        print(f"📖 Reader {reader_id} finished reading.")
        await rw_lock.release_read()
        await asyncio.sleep(0.5)  # Sleep before next read

async def writer(writer_id):
    for i in range(2):
        await rw_lock.acquire_write()
        print(f"✍️ Writer {writer_id} has EXCLUSIVE file access. Writing...")
        await asyncio.sleep(1.0)  # Simulate file writing I/O
        print(f"✍️ Writer {writer_id} finished writing.")
        await rw_lock.release_write()
        await asyncio.sleep(0.8)  # Give others a chance to grab the lock

async def main():
    # Launch 2 readers and 2 writers concurrently
    await asyncio.gather(
        reader(1), reader(2),
        writer(1), writer(2)
    )

if __name__ == "__main__":
    asyncio.run(main())
```

```python

# Example of asyncio.create_task():

import asyncio
import nest_asyncio
nest_asyncio.apply()

async def worker(n):
      print(f"Start of worker {n}")
      await asyncio.sleep(2)
      print(f"End of worker {n}")

async def main():
      print("Start of main coroutine")
      tasks=[]
      for i in range(10):
            tasks.append( asyncio.create_task(worker(i)))
      await asyncio.gather(*tasks)
      print("End of main coroutine")


if __name__ == "__main__":
      asyncio.run(main())
      
```

Here are the main concurrency control mechanisms in asyncio:

1. asyncio.Lock
Lets only one coroutine enter a critical section at a time
Use it to protect shared data
2. asyncio.Semaphore
Lets a fixed number of coroutines enter at once
Use it to limit concurrency, like 5 downloads at a time
3. asyncio.BoundedSemaphore
Same as Semaphore, but safer
Prevents releasing more times than acquired
4. asyncio.Event
One coroutine sets a flag, others wait until it is set
Use it for signaling
5. asyncio.Condition
Combines a lock and event-style waiting
Use it when coroutines need to wait for a specific condition before continuing
6. asyncio.Queue
Good for producer/consumer patterns
Controls how many items are waiting and how many workers consume them
7. asyncio.TaskGroup Python 3.11+
Manages a group of tasks together
Not a lock, but useful for structured concurrency


Simple mental model:

Lock = one at a time
Semaphore = N at a time
BoundedSemaphore = N at a time, with safety check
Event = wait until told to proceed
Condition = wait until a specific state changes
Queue = hand work to workers safely
TaskGroup = organize many tasks together

For boto3 + S3 work, the most useful asyncio control usually depends on the pattern:

Mostly used
asyncio.Semaphore
This is the most common one for S3-style automation.

Why:

you often want to run many uploads/downloads
but not too many at once
to avoid overload, throttling, or exhausting local resources
Example use:

upload 100 objects
but only 5 or 10 at a time
Also very useful
asyncio.Queue
This is great when you have a pipeline:

producer adds S3 tasks
worker coroutines process them
It is often the best choice when work arrives dynamically.

For shared data
asyncio.Lock
Use this if multiple coroutines update the same:

counters
status dictionary
shared log structure
progress tracker
Construct for running tasks in a particular order
asyncio.Queue
If you want tasks to run in a controlled order, a queue is the usual answer.

Why:

FIFO order
consumers process items in the order they arrive
Example:

task A goes in first
task B goes in second
worker handles A before B
If you need strict step-by-step sequencing, you can also just await tasks one by one, but that is not concurrency.

Quick rule for S3
Limit number of active S3 calls → Semaphore
Process tasks in arrival order → Queue
Protect shared state → Lock
Signal between tasks → Event
Practical S3 example
If you are doing parallel S3 uploads, I would usually start with:

Semaphore to cap active transfers
Queue if tasks are coming in continuously
Lock only for shared counters or result aggregation

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        mapping = {")": "(", "}": "{", "]": "["}
        if len(s) % 2 != 0:
            return False

        for chr in s:
            if chr in mapping.values():
                stack.append(chr)
            elif chr in mapping :
                if not stack:
                    return False
                top_element = stack.pop()
                if mapping[chr] != top_element:
                    return False
            else :
                return False
        return not stack
    

```

Remove consecutive duplicate from a string

```python
def remove_duplicates(s:str)->str:
    if not s:
        return ""
    temp=[s[0]]
    for i in range(1,len(s)):
        if s[i] !=temp[-1]:
            temp.append(s[i])
    return "".join(temp)




st="aabcccddaa"
print(remove_duplicates(st))


```

```python
# Check if two strings are anagrams of each other, 
# ignoring spaces and case sensitivity.

def are_anagrams(str1: str, str2: str) -> bool:
    str1=str1.replace(" ", "").lower()
    str2=str2.replace(" ", "").lower()
    
    if len(str1) != len(str2):
        return False
    
    seen={}
    for ch in str1:
        seen[ch]=seen.get(ch,0)+1
    for ch in str2:
        seen[ch]=seen.get(ch,0)-1
    return all(value == 0 for value in seen.values())

```
