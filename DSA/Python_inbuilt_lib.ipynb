```python
# Contains Duplicate — set-based

def has_duplicates(lst):
    lkp=set()
    for i in lst:
        if i in lkp:
            return True
        lkp.add(i)
    return False 

nums = [1, 2, 3, 4]
print(has_duplicates(nums))

nums = [1, 2, 3, 1]
print(has_duplicates(nums))
```

```python
#Product of Array Except Self — no division, O(n)
def product_except_self(nums):
    left_product=[1]*len(nums)
    right_product=[1]*len(nums)
    result=[]
    for i in range(1,len(nums)):
        left_product[i] = left_product[i-1]*nums[i-1]
    for j in range(len(nums)-2, -1, -1):
            right_product[j] = right_product[j+1] * nums[j+1]

    for i in range(len(nums)):
        result.append(left_product[i]*right_product[i])
    return result

nums = [1, 2, 3, 4]
print(product_except_self(nums))
```

```python
# Maximum sum subarray brute force O(n^2)
def max_subarray_sum(nums):
    max_sum = float('-inf')
    start_index = 0
    end_index = 0
    
    for i in range(len(nums)):
        current_sum = 0
        
        for j in range(i, len(nums)):
            current_sum += nums[j]
            if current_sum > max_sum:
                max_sum = current_sum
                start_index = i
                end_index = j
    return max_sum, nums[start_index:end_index+1]

nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
max_sum, subarray = max_subarray_sum(nums)
print("Maximum sum:", max_sum)
print("Subarray:", subarray)
        
```

```python
# Maximum sum subarray using Kadane's algorithm O(n)
# Intution: If the current sum is negative, we start a new subarray from the next element. If the current sum is positive, we continue adding to the current subarray. We keep track of the maximum sum encountered so far.

def kadane_max_subarray_sum(nums):
    max_sum=nums[0]
    current_sum=nums[0]
    start=0
    best_start=0
    best_end=0
    for i in range(1,len(nums)):
        if current_sum<0:
            current_sum=nums[i]
            start=i
        else:
            current_sum+=nums[i]
        
        if current_sum>max_sum:
            max_sum=current_sum
            best_start=start
            best_end=i
    return max_sum,nums[best_start:best_end+1]

nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
max_sum, subarray = kadane_max_subarray_sum(nums)
print("Maximum sum:", max_sum)
print("Subarray:", subarray)

```

```python
# Binary search
def binary_search(arr, target):
    mid=len(arr)//2
    if arr[mid]==target:
        return mid
    elif arr[mid]<target:
        result = binary_search(arr[mid+1:], target)
        return result 
    else:
        return binary_search(arr[:mid], target)
    return None

arr = [1, 2, 3, 4, 5]
target = 3
index = binary_search(arr, target)
print("Index:", index)


    
```

```python
# Print the linked list in reverse order using recursion

class Node:
    def __init__(self, value):
        self.data=value
        self.next=None


class DoublyNode:
    def __init__(self,key:int|None=None,value:int|None=None):
        self.key=key
        self.value=value
        self.next=None
        self.prev=None



class LinkedList:
    def __init__(self):
        self.head=None
    def append(self,value):
        new_node=Node(value)

        """ Insert if the linked list is empty"""
        if self.head is None:
            self.head=new_node
            return
        else:
            """ Insert at the end of the linked list"""
            tail=self.head
        while tail.next:
            tail= tail.next
        """ Insert at the end of the linked list"""
        tail.next=new_node

    def print_list(self):
        current=self.head
        while current:
            print(current.data)
            current=current.next
    def print_reverse_recursively(self,node):
        if node is None:
            return
        self.print_reverse_recursively(node.next)
        print(node.data)

    def print_reverse_iteratively(self):
        if self.head is None:
            return
        current=self.head
        stack=[]
        while current:
            stack.append(current.data)
            current=current.next
        while stack:
            print(stack.pop())

    def middle_node(self):
        slow=self.head
        fast=self.head
        while fast and fast.next:
            slow=slow.next
            fast=fast.next.next
        return slow

    def detect_cycle(self):
        slow=self.head
        fast=self.head
        while fast and fast.next:
            slow=slow.next
            fast=fast.next.next
            if slow==fast:
                return True
        return False

    

            


if __name__=="__main__":
    ll=LinkedList()
    ll.append(1)
    ll.append(2)
    ll.append(3)
    ll.append(4)
    ll.append(5)
    ll.append(6)
    #ll.print_list()
    #ll.print_reverse_recursively(ll.head)
    #ll.print_reverse_iteratively()
    print(ll.middle_node().data)


```

```python
# Count number of bits in an integer
def count_bits(n):
    count=0
    while n:
        if n&1:
            count+=1
        n>>=1
    return count

if __name__=="__main__":
    n=11
    print(count_bits(n))
```

```python
# Missing single number in an array using XOR
def missing_number(arr):
    n=arr[0]
    for i in range(1,len(arr)):
        n^=arr[i]
    return n

if __name__=="__main__":
    arr=[1,2,2,3,3,4,4,5,6,6,7,7,9,8,8,9,10,10,1]
    print(missing_number(arr))
```

```python
# Missing number in an array using XOR
def missing_number2(arr):
    n=0
    for i in range(len(arr)):
        n^=arr[i]
    for i in range(1,len(arr)+2):
        n^=i
    return n

if __name__=="__main__":
    arr=[1,2,3,4,5,6,7,8,10]
    print(missing_number2(arr))
```
